```python
import pandas as pd
import numpy as np

from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)

# =========================
# 1. Load locked BioMapAI model constants
# =========================

panel = pd.read_csv("C:/Users/rohan/OneDrive/Desktop/ME-CFS/src/MECFS_FINAL_SIGNATURES/NB1_LOCKED_10_PANEL/01_locked_10_metabolite_panel.csv")["feature"].tolist()

coef_df = pd.read_csv("C:/Users/rohan/OneDrive/Desktop/ME-CFS/src/MECFS_FINAL_SIGNATURES/NB1_LOCKED_10_PANEL/02_10_metabolite_coefficients.csv")
coef_col = [c for c in coef_df.columns if c != "feature"][0]
coefs = coef_df.set_index("feature")[coef_col]

intercept_df = pd.read_csv("C:/Users/rohan/OneDrive/Desktop/ME-CFS/src/MECFS_FINAL_SIGNATURES/NB1_LOCKED_10_PANEL/03_10_metabolite_intercept.csv")
intercept = intercept_df.iloc[0, 0]

scaler = pd.read_csv("C:/Users/rohan/OneDrive/Desktop/ME-CFS/src/MECFS_FINAL_SIGNATURES/NB1_LOCKED_10_PANEL/04_10_metabolite_scaler_constants.csv").set_index("feature")

mean_col = [c for c in scaler.columns if "mean" in c.lower()][0]
scale_col = [c for c in scaler.columns if "scale" in c.lower() or "std" in c.lower()][0]

# =========================
# 2. Load Germain external data
# =========================

meta = pd.read_excel("study_data_mecfs.xlsx", sheet_name="sample_metadata")
dd = pd.read_excel("study_data_mecfs.xlsx", sheet_name="data_dictionary")
mat = pd.read_excel("study_data_mecfs.xlsx", sheet_name="data_matrix")

# =========================
# 3. Map compound_id to biochemical name
# =========================

compound_to_name = dd.set_index("compound_id")["BIOCHEMICAL"].astype(str).to_dict()

mat_named = mat.copy()
mat_named["BIOCHEMICAL"] = mat_named["compound_id"].map(compound_to_name)

# =========================
# 4. Keep only exact matched panel metabolites
# =========================

available = []
missing = []

for m in panel:
    if m in set(mat_named["BIOCHEMICAL"]):
        available.append(m)
    else:
        missing.append(m)

print("Available exact panel metabolites:", len(available))
print(available)

print("\nMissing panel metabolites:", len(missing))
print(missing)

# =========================
# 5. Build external X: samples × metabolites
# =========================

sample_cols = [c for c in mat_named.columns if str(c).startswith("sample_")]

X_ext = (
    mat_named
    .loc[mat_named["BIOCHEMICAL"].isin(available), ["BIOCHEMICAL"] + sample_cols]
    .set_index("BIOCHEMICAL")
    .T
)

X_ext.index.name = "sample_id"

# align metadata to sample rows
meta2 = meta.copy()
meta2["sample_id"] = meta2["sample_id"].astype(str)

X_ext = X_ext.reset_index().merge(meta2, on="sample_id", how="inner")

y_ext = X_ext["health_status"].str.lower().map({
    "case": 1,
    "control": 0
})

X_values = X_ext[available].copy()

print("\nExternal X shape:", X_values.shape)
print("External y counts:")
print(y_ext.value_counts())

# =========================
# 6. Apply BioMapAI scaler constants
# IMPORTANT: use original BioMapAI means/scales, not Germain means
# =========================

X_scaled = pd.DataFrame(index=X_values.index)

for m in available:
    X_scaled[m] = (X_values[m] - scaler.loc[m, mean_col]) / scaler.loc[m, scale_col]

# =========================
# 7. Apply locked BioMapAI coefficients
# Missing metabolites are simply excluded from this reduced external score
# =========================

logits = np.full(len(X_scaled), intercept, dtype=float)

for m in available:
    logits += X_scaled[m].values * coefs.loc[m]

probs = 1 / (1 + np.exp(-logits))

# =========================
# 8. Evaluate external performance
# =========================

auc = roc_auc_score(y_ext, probs)

preds_05 = (probs >= 0.5).astype(int)

acc = accuracy_score(y_ext, preds_05)
bal_acc = balanced_accuracy_score(y_ext, preds_05)
cm = confusion_matrix(y_ext, preds_05)

print("\n" + "="*80)
print("GERMAIN EXTERNAL VALIDATION USING AVAILABLE PANEL METABOLITES")
print("="*80)

print("Metabolites used:", len(available), "/", len(panel))
print("AUC:", auc)
print("Accuracy @ 0.5:", acc)
print("Balanced accuracy @ 0.5:", bal_acc)

print("\nConfusion matrix:")
print(cm)

print("\nClassification report:")
print(classification_report(y_ext, preds_05, target_names=["control", "case"]))

# =========================
# 9. Save predictions
# =========================

pred_out = pd.DataFrame({
    "sample_id": X_ext["sample_id"],
    "health_status": X_ext["health_status"],
    "y_true": y_ext,
    "locked_model_probability": probs,
    "pred_05": preds_05
})

pred_out.to_csv("germain_external_predictions_reduced_panel.csv", index=False)

display(pred_out.head())
```
